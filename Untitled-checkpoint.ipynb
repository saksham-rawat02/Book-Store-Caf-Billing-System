```python
import tkinter as tk
from tkinter import ttk, messagebox
from datetime import datetime


# ============================================================
# BOOK STORE CAFE - BILLING SYSTEM
# ============================================================

# ---------------- Menu ----------------
MENU = {
    "Pasta": 250,
    "Maggi": 80,
    "Coffee": 70,
    "Veg Puff": 130,
    "Pizza": 99,
    "Burger": 150,
    "Sandwich": 120,
    "French Fries": 100,
    "Cold Coffee": 110,
    "Tea": 50,
    "Chocolate Cake": 180,
    "Momos": 140,
    "Garlic Bread": 160,
    "Noodles": 130,
    "Ice Cream": 90
}

GST_RATE = 0.18

# Stores ordered items
items = []


# ============================================================
# FUNCTIONS
# ============================================================

def add_item():
    """Add the selected item and quantity to the order."""

    item = combo.get()
    quantity_text = quantity_entry.get()

    if not item:
        messagebox.showwarning(
            "Selection Required",
            "Please select an item from the menu."
        )
        return

    try:
        quantity = int(quantity_text)

        if quantity <= 0:
            raise ValueError

    except ValueError:
        messagebox.showerror(
            "Invalid Quantity",
            "Please enter a valid quantity greater than 0."
        )
        return

    price = MENU[item]
    amount = price * quantity

    # Store item details
    items.append({
        "name": item,
        "price": price,
        "quantity": quantity,
        "amount": amount
    })

    # Add item to Listbox
    listbox.insert(
        tk.END,
        f"{item:<15} x {quantity:<3} = ₹{amount:.2f}"
    )

    update_total()

    # Reset selection
    combo.set("")
    quantity_entry.delete(0, tk.END)
    quantity_entry.insert(0, "1")


def remove_item():
    """Remove the selected item from the order."""

    selected = listbox.curselection()

    if not selected:
        messagebox.showwarning(
            "No Selection",
            "Please select an item to remove."
        )
        return

    index = selected[0]

    # Remove item from list
    items.pop(index)

    # Remove item from Listbox
    listbox.delete(index)

    update_total()


def clear_order():
    """Clear the complete order."""

    if not items:
        messagebox.showinfo(
            "Empty Order",
            "There are no items to clear."
        )
        return

    confirm = messagebox.askyesno(
        "Clear Order",
        "Are you sure you want to clear the entire order?"
    )

    if confirm:
        items.clear()
        listbox.delete(0, tk.END)
        update_total()


def calculate_subtotal():
    """Calculate the subtotal of all ordered items."""

    return sum(item["amount"] for item in items)


def update_total():
    """Update the subtotal displayed on the screen."""

    subtotal = calculate_subtotal()

    total_label.config(
        text=f"Current Subtotal: ₹{subtotal:.2f}"
    )


def generate_bill():
    """Generate and display the final bill."""

    if not items:
        messagebox.showwarning(
            "Empty Order",
            "No items have been added to the order."
        )
        return

    # Get customer name
    customer_name = customer_entry.get().strip()

    if not customer_name:
        customer_name = "Guest"

    # Calculate bill
    subtotal = calculate_subtotal()
    gst = subtotal * GST_RATE
    final_total = subtotal + gst

    # Generate bill number
    bill_number = datetime.now().strftime("%Y%m%d%H%M%S")

    # Current date and time
    date_time = datetime.now().strftime(
        "%d-%m-%Y  %I:%M %p"
    )

    # ========================================================
    # BILL FORMAT
    # ========================================================

    bill_text = ""

    bill_text += "       BOOK STORE CAFE\n"
    bill_text += "================================\n"
    bill_text += f"Bill No. : {bill_number}\n"
    bill_text += f"Customer : {customer_name}\n"
    bill_text += f"Date     : {date_time}\n"
    bill_text += "================================\n"

    bill_text += "Item            Qty      Amount\n"
    bill_text += "--------------------------------\n"

    # Add each item to bill
    for item in items:

        bill_text += (
            f"{item['name']:<16}"
            f"{item['quantity']:<9}"
            f"₹{item['amount']:.2f}\n"
        )

    bill_text += "--------------------------------\n"

    bill_text += f"Subtotal       : ₹{subtotal:.2f}\n"
    bill_text += f"GST (18%)      : ₹{gst:.2f}\n"

    bill_text += "================================\n"

    bill_text += f"TOTAL BILL     : ₹{final_total:.2f}\n"

    bill_text += "================================\n"

    bill_text += "     Thank You! Visit Again!\n"

    # Display bill
    messagebox.showinfo(
        "Final Bill",
        bill_text
    )


def exit_application():
    """Close the application after confirmation."""

    confirm = messagebox.askyesno(
        "Exit",
        "Are you sure you want to exit?"
    )

    if confirm:
        root.destroy()


# ============================================================
# GUI WINDOW
# ============================================================

root = tk.Tk()

root.title("Book Store Cafe - Billing System")
root.geometry("650x760")
root.resizable(False, False)

root.config(
    bg="#FFF8E7"
)


# ============================================================
# HEADER
# ============================================================

title = tk.Label(
    root,
    text="BOOK STORE CAFE",
    font=("Arial", 22, "bold"),
    bg="#6B3E26",
    fg="white",
    pady=15
)

title.pack(fill=tk.X)


subtitle = tk.Label(
    root,
    text="Cafe Ordering & Billing System",
    font=("Arial", 11, "italic"),
    bg="#FFF8E7",
    fg="#6B3E26"
)

subtitle.pack(pady=5)


# ============================================================
# CUSTOMER DETAILS
# ============================================================

customer_frame = tk.LabelFrame(
    root,
    text="Customer Details",
    font=("Arial", 11, "bold"),
    bg="#FFF8E7",
    fg="#6B3E26",
    padx=10,
    pady=10
)

customer_frame.pack(
    padx=25,
    pady=10,
    fill=tk.X
)


customer_label = tk.Label(
    customer_frame,
    text="Customer Name:",
    font=("Arial", 11),
    bg="#FFF8E7"
)

customer_label.grid(
    row=0,
    column=0,
    padx=5,
    pady=5
)


customer_entry = tk.Entry(
    customer_frame,
    font=("Arial", 11),
    width=35
)

customer_entry.grid(
    row=0,
    column=1,
    padx=5,
    pady=5
)


# ============================================================
# MENU / ITEM SELECTION
# ============================================================

menu_frame = tk.LabelFrame(
    root,
    text="Select Food Item",
    font=("Arial", 11, "bold"),
    bg="#FFF8E7",
    fg="#6B3E26",
    padx=10,
    pady=10
)

menu_frame.pack(
    padx=25,
    pady=5,
    fill=tk.X
)


# Food Item Label
food_label = tk.Label(
    menu_frame,
    text="Food Item:",
    font=("Arial", 11),
    bg="#FFF8E7"
)

food_label.grid(
    row=0,
    column=0,
    padx=5,
    pady=5
)


# Food Item Dropdown
combo = ttk.Combobox(
    menu_frame,
    values=list(MENU.keys()),
    font=("Arial", 11),
    state="readonly",
    width=20
)

combo.grid(
    row=0,
    column=1,
    padx=5,
    pady=5
)


# Quantity Label
quantity_label = tk.Label(
    menu_frame,
    text="Quantity:",
    font=("Arial", 11),
    bg="#FFF8E7"
)

quantity_label.grid(
    row=0,
    column=2,
    padx=5,
    pady=5
)


# Quantity Entry
quantity_entry = tk.Entry(
    menu_frame,
    font=("Arial", 11),
    width=6
)

quantity_entry.grid(
    row=0,
    column=3,
    padx=5,
    pady=5
)

quantity_entry.insert(
    0,
    "1"
)


# ============================================================
# ADD ITEM BUTTON
# ============================================================

add_button = tk.Button(
    root,
    text="➕ Add Item",
    font=("Arial", 11, "bold"),
    bg="#2E8B57",
    fg="white",
    activebackground="#246B45",
    activeforeground="white",
    width=18,
    command=add_item
)

add_button.pack(
    pady=10
)


# ============================================================
# ORDER LIST
# ============================================================

order_frame = tk.LabelFrame(
    root,
    text="Current Order",
    font=("Arial", 11, "bold"),
    bg="#FFF8E7",
    fg="#6B3E26",
    padx=10,
    pady=10
)

order_frame.pack(
    padx=25,
    pady=5,
    fill=tk.BOTH
)


listbox = tk.Listbox(
    order_frame,
    width=60,
    height=9,
    font=("Courier New", 11),
    bg="white",
    fg="black",
    selectbackground="#6B3E26",
    selectforeground="white"
)

listbox.pack()


# ============================================================
# REMOVE / CLEAR BUTTONS
# ============================================================

button_frame = tk.Frame(
    root,
    bg="#FFF8E7"
)

button_frame.pack(
    pady=10
)


# Remove Button
remove_button = tk.Button(
    button_frame,
    text="Remove Item",
    font=("Arial", 10, "bold"),
    bg="#D9534F",
    fg="white",
    width=14,
    command=remove_item
)

remove_button.grid(
    row=0,
    column=0,
    padx=5
)


# Clear Button
clear_button = tk.Button(
    button_frame,
    text="Clear Order",
    font=("Arial", 10, "bold"),
    bg="#F0AD4E",
    fg="white",
    width=14,
    command=clear_order
)

clear_button.grid(
    row=0,
    column=1,
    padx=5
)


# ============================================================
# TOTAL
# ============================================================

total_label = tk.Label(
    root,
    text="Current Subtotal: ₹0.00",
    font=("Arial", 14, "bold"),
    bg="#FFF8E7",
    fg="#6B3E26"
)

total_label.pack(
    pady=8
)


# ============================================================
# GENERATE BILL BUTTON
# ============================================================

bill_button = tk.Button(
    root,
    text="🧾 GENERATE BILL",
    font=("Arial", 13, "bold"),
    bg="#1E5AA8",
    fg="white",
    activebackground="#16457F",
    activeforeground="white",
    width=25,
    height=2,
    command=generate_bill
)

bill_button.pack(
    pady=8
)


# ============================================================
# EXIT BUTTON
# ============================================================

exit_button = tk.Button(
    root,
    text="Exit",
    font=("Arial", 10, "bold"),
    bg="#555555",
    fg="white",
    width=12,
    command=exit_application
)

exit_button.pack(
    pady=5
)


# ============================================================
# FOOTER
# ============================================================

footer = tk.Label(
    root,
    text="Thank you for visiting Book Store Cafe!",
    font=("Arial", 9, "italic"),
    bg="#FFF8E7",
    fg="#777777"
)

footer.pack(
    pady=8
)


# ============================================================
# START APPLICATION
# ============================================================

root.mainloop()
```
